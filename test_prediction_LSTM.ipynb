```python
from tensorflow.keras.models import load_model
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
import tensorflow as tf
import numpy as np
```

```python
# Get Word Index:
word_index=imdb.get_word_index()
```

```python
# Load Model:
model=load_model('lstm_imdb.keras')
model.summary()
```

```python
def processed_input(review):
    max_length=500
    words=review.lower().split()
    encoded_review=[word_index.get(word, 2) + 3 for word in words]
    padded_review=sequence.pad_sequences([encoded_review],maxlen=max_length)

    return padded_review
```

```python
# Prediction Function:

def predict_sentiment(review):
    encoded_review=processed_input(review)
    prediction=model.predict(encoded_review)
    score=prediction[0][0]
    sentiment= 'Positive' if score>=0.5 else 'Negative'
    print(f"Review: {review}")
    print(f"Prediction Score: {score}")
    print(f"Sentiment: {sentiment}")
```

```python
example_review1= "This movie was fantastic! The acting was great and the plot was thrilling."
predict_sentiment(example_review1)

example_review2="The movie failed to meet even basic expectations."
predict_sentiment(example_review2)
```
